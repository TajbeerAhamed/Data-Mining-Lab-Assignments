```python
# Importing libraries and dataset
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
df = pd.read_csv('/content/drive/MyDrive/Data_Mining_LAb/Assignments/Project_2/heart_disease.csv')
df
```

```python
# Encoding Dataset
import pandas as pd
df_encoded = pd.get_dummies(df, columns=df.columns)
df_encoded
```

```python
# Frequent itemset generation
frequent_itemsets = apriori(df_encoded, min_support=0.3,
use_colnames=True)
print(frequent_itemsets)
```

```python
# Association rule generation
num_itemsets = len(frequent_itemsets)
rules = association_rules(frequent_itemsets, metric="confidence",
min_threshold=0.7, num_itemsets = num_itemsets)
rules
```

```python
# Generate the top 10 association rules based on confidence and lift
rules = rules.sort_values(by=['confidence', 'lift'], ascending=False)
rules.head(10)
```

```python
# Step 3: Sort rules by lift in descending order and select top 10
top_10_rules = rules.sort_values(by=["confidence", "lift"],
ascending=False).head(10)
```

```python
# Step 4: Display the top 10 rules
print(top_10_rules[['confidence', 'lift']])
```
